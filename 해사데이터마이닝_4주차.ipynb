```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch7/air_quality.csv")

# 1) IQR 계산
Q1 = df['CO2'].quantile(0.25)
Q3 = df['CO2'].quantile(0.75)
IQR = Q3 - Q1
print("IQR 값: ", IQR)

# 2) 상한 및 하한 계산
upper = Q3 + 1.5 * IQR
lower = Q1 - 1.5 * IQR

# 3) 이상치 식별
outliers = df[(df['CO2'] < lower) | (df['CO2'] > upper)] #'|': and 연산자
print("\n", outliers)

# 4) 이상치 수
print("\n", len(outliers))
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# df = pd.read_csv("chem.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch8/chem.csv")

# 1) Min-Max 스케일링 (데이터프레임으로 만들기)
scaler = MinMaxScaler() # 데이터의 최소값을 0, 최대값을 1로 하여 값을 변환
df['co_scaled'] = scaler.fit_transform(df[['co']])
df['nmhc_scaled'] = scaler.fit_transform(df[['nmhc']])
print(df) # 값이 모두 0과 1 사이에 존재

# 2) 표준편차 계산 (판다스 표준편차)
co_std = df['co_scaled'].std()
nmhc_std = df['nmhc_scaled'].std()
print(co_std, nmhc_std)

# 3) 표준편차 차이 계산 및 반올림
std_diff = round(co_std - nmhc_std, 3)
print(std_diff)
```

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/main/part4/ch9/loan.csv")

# 1) 총대출액 컬럼 생성
df['총대출액'] = df['신용대출'] + df['담보대출']
print("<총대출액이 포함된 컬럼>")
print(df)

# 2) 지역코드와 성별별로 총대출액 합계 계산
grouped = df.groupby(['지역코드', '성별'])['총대출액'].sum()
print("<unstack()가 없는 경우>")
print(grouped)

grouped_unstack = df.groupby(['지역코드', '성별'])['총대출액'].sum().unstack()
print("\n<unstack()가 있는 경우>")
print(grouped_unstack)

pivot_result = df.pivot_table(index='지역코드', columns='성별', values='총대출액', aggfunc='sum')
print("\n<picot_table을 사용한 경우>")
print(pivot_result)

# 3) 성별 간 총대출액 차이가 가장 큰 지역코드 찾기 (절대값 사용)
grouped_unstack['차이'] = abs(grouped_unstack[1] - grouped_unstack[2])
result = grouped_unstack['차이'].idxmax() # 값이 가장 큰 인덱스 반환
print("\n성별에 따른 총대출액 차가 가장 큰 지역코드:", result)
```

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/main/part4/ch9/crime.csv")

# 1) "발생건수"와 "검거건수"를 따로 분리
cond1 = df["구분"] == "발생건수" # 발생건수만 가져오기
df1 = df[cond1].iloc[:, 2:]
print("<발생건수>")
print(df1)

cond2 = df["구분"] == "검거건수"# 검거건수만 가져오기
df2 = df[cond2].iloc[:, 2:]
print("<검거건수>")
print(df2)

# 2) 검거율 계산 (검거건수 / 발생건수)
df1 = df1.reset_index(drop=True) # reset_index(): 기존의 인덱스를 새로 매기는 함수
df2 = df2.reset_index(drop=True)
df3 = df2 / df1
print(df1, "\n")

# 3) 각 연도에서 검거율이 가장 높은 범죄유형 찾기
listbox = df3.idxmax(axis=1)
print(listbox)

# 4) 가장 높은 검거율을 기록한 범죄유형의 검거건수 가져오기
result = 0
for index, item in enumerate(listbox): # for 루프를 사용하여 검거건수를 합함
    result += df2.loc[index, item]
    # enumerate(): listbox를 듀플로 만들어 값의 수만큼 반복
    # 각각의 범죄유형의 검거검수를 더해줌

print("\n연도별 최대 검거율을 기록한 범죄의 검거건수의 합:", result)
```

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/main/part4/ch9/crime.csv")

# 1. 데이터를 길게 변환
df_long = pd.melt(df, id_vars=["연도", "구분"], var_name="범죄유형", value_name="건수")
print(df_long) # melt를 사용하여 연도, 구분, 범죄유형을 기준으로 재구성

# 2. 피벗(pivot)하여 발생건수와 검거건수를 한 행에 모으기:
df_pivot = df_long.pivot_table(index=["연도", "범죄유형"], columns="구분", values="건수").reset_index()
print(df_pivot) # pivot_table을 사용해 연도별로 정렬된 행 생성

# 3. 검거율 계산
df_pivot["검거율"] = df_pivot["검거건수"] / df_pivot["발생건수"]

# 4. 각 연도별로 검거율이 가장 높은 범죄유형 찾기
listbox = df_pivot.groupby("연도")["검거율"].idxmax()
max_crime = df_pivot.loc[listbox]
print(listbox)

# 5. 각 연도별 최고 검거율 범죄유형의 검거건수를 모두 합산하기
result = int(max_crime["검거건수"].sum())
print("\n연도별 최대 검거율을 기록한 범죄의 검거건수의 합:", result)
```

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/main/part4/ch9/hr.csv")

# 1) 만족도 결측치 처리
m = df["만족도"].mean()
df["만족도"] = df["만족도"].fillna(m)

# 2) 근속연수 결측치 처리 (부서와 성과등급 기준 평균값으로 채움)
gm = df.groupby(["부서", "성과등급"])["근속연수"].transform("mean")
gm = gm.astype(int)
df["근속연수"] = df["근속연수"].fillna(gm)

# 3) 연봉 / 근속연수 계산 후 세 번째로 높은 사람의 근속연수 (A)
df["연봉_근속연수"] = df["연봉"] / df["근속연수"]
df_year = df.nlargest(3,'연봉_근속연수') # 세 번째로 높은 값을 구하는 함수
A = int(df_year.iloc[-1]["근속연수"])
print("'연봉/근속연수'의 값이 세 번째로 높은 사람의 근속연수:", A)

# 4) 연봉 / 만족도 계산 후 두 번째로 높은 사람의 교육참가횟수 (B)
df["연봉_만족도"] = df["연봉"] / df["만족도"]
df_like = df.sort_values(by="연봉_만족도", ascending=False) # 값을 내림차순 정렬
B = df_like.iloc[1]["교육참가횟수"] # 두 번째로 높은 값을 추출
print("'연봉/만족도'의 값이 두 번째로 높은 사람의 교육참가횟수:", B)

# 5) 결과 출력
result = A + B
print(result)
```
