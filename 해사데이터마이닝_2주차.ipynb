```python
import pandas as pd
# df = pd.read_csv("members.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch2/members.csv")

# 1) 위에서부터 80%
line = int(len(df) * 0.8)
df = df.iloc[:line]  #iloc: 행의 순서를 나타내는 정수로 값을 추출
print(df)

# 2) 'f1' 표준편차
std1 = df['f1'].std()
print("\n\n기본 표준편차: ", round(std1, 2))  #둘째자리까지 표현

# 3) 중앙값 채우기
med=df['f1'].median()
df['f1'] = df['f1'].fillna(med)

# 4) 'f1' 중앙값 표준편차
std2 = df['f1'].std()
print("중앙값 표준편차: ", round(std2, 2))  #둘째자리까지 표현

# 5) 두 표준편차 차이 절대값 계산 (반올림하여 소수 둘째자리 표현)
result = abs(std1-std2)
print(round(result,2))
```

```python
import pandas as pd
# df = pd.read_csv("members.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch2/members.csv")
s = [0,0,0,0,0,0]

# 1) 표준편차, 평균값 구하기
s[0] = df['age'].std()
s[1] = df['age'].mean()

# 2) 이상치 범위 구하기
s[2] = s[1] - (s[0] * 1.5)
s[3] = s[1] + (s[0] * 1.5)

# 3) 이상치 구하기
s[4] = df['age'] < s[2]
s[5] = df['age'] > s[3]

# 4) 이상치의 합
print(df[s[4]|s[5]]['age'].sum())
```
